## Testing Gliner for Entity Extraction

```python
from gliner import GLiNER

model = GLiNER.from_pretrained("knowledgator/modern-gliner-bi-large-v1.0", load_tokenizer=True)

text = """
SPEAKER_00: Good morning, David. It's been about two weeks since our last session here at the Maple Street Clinic. How have you been adjusting to the new dosage?
SPEAKER_01: Honestly, Dr. Chen, it's been rough. Ever since Dr. Aris bumped the Sertraline up to 100 milligrams last Tuesday, I've been dealing with terrible insomnia.
SPEAKER_00: I'm sorry to hear that. Insomnia is a known side effect when titrating up. Can you tell me what your sleep schedule looks like right now?
SPEAKER_01: I usually fall asleep around 11:00 PM, but then I wake up at 2:30 AM with chest tightness. It feels like the precursor to a panic attack, similar to the ones I had back in Seattle in 2022.
SPEAKER_00: That chest tightness can definitely be anxiety-induced. Have you tried using the 4-7-8 breathing technique we discussed when those nighttime awakenings happen?
SPEAKER_01: I tried it on Thursday night. It helped lower my heart rate, but my mind kept racing about the quarterly report due at work next Friday. I ended up calling my sister, Sarah, just to ground myself.
SPEAKER_00: It's great that you reached out to Sarah for support. How did you feel the next day after missing out on that sleep?
SPEAKER_01: Completely drained. I actually had to take a half-day off from the accounting firm on Friday because the brain fog was so intense I couldn't look at the spreadsheets.
SPEAKER_00: Let's make a note to message Dr. Aris through the portal about this. He might suggest pairing it with a low-dose Trazodone temporarily to help you sleep while your body adjusts.
SPEAKER_01: That would be helpful. I'm willing to try it, I just can't afford another week of operating on three hours of sleep.
"""

# labels = ["person", "award", "date", "competitions", "teams"]
labels = ["symptom", "trigger", "emotion", "person", "coping mechanism", "life event", "behavior"]

entities = model.predict_entities(text, labels, threshold=0.3)

for entity in entities:
    print(entity["text"], "=>", entity["label"])
```
